### KEYWORDS

```python
skills = [
    "SQL", "Python", "Excel", "Tableau", "Power BI"," BI",
    "Visual", "Statistic", "Machine Learning", " ML ", " AI", "Artificial Intelligence",
    "Cleaning", "Wrangling", "ETL", "Big Data", "Data Modeling",
    "Intelligence", "Reporting", "Dashboard", "A/B",
    "Predictive", "Data Mining", "Google Sheet","Google Analytics", "Looker",
    "Snowflake", "Spark", "Hadoop", "Critical Thinking", "Problem Solving",
    "Communication", "Stakeholder", "Management", "Governance", "KPI",
    "Metrics", "Pipeline", "Warehouse", "Azure", "AWS", "GCP",
    "Pandas", "NumPy", "Matplotlib", "Seaborn", "Jupyter", "Git",
    "Storytelling", "Regression", "Hypothesis",
    "Strategy", "spreadsheet"
]

responsibility = [
    "analyze", "report", "dashboards", "trends", "insights",
    "decision-making", "collaborate", "clean", "organize", "monitor",
    "present", "findings", "track", "KPI", "build", "interpret",
    "gather requirements", "automate", "accuracy", "datasets", "analysis",
    "visual", "solve", "business problems", "communicate", "metrics",
    "validation", "stakeholders", "extract", "transform", "load", "query",
    "document", "forecast", "trends"
]

qualifications = [
    "bachelor degree", "graduate", "education", "master's degree", "mathematics", "computer science", "certificate",
    "engineering", "economics", "information technology", "I.T", "data science", "experience",
    "excellent communication skills", "certifications", "Google Data Analytics",
    "Microsoft Certified", "AWS Certified",
    "portfolio", "project", "internship", "entry", "mid",
    "senior", "internet connection"
]
```

```python
def comparison(description, keywords):
    """
    Checks which keywords from a list appear in the job description.

    Args:
        description (str): Full text of the job posting.
        keywords (list):   List of keywords to search for.

    Returns:
        list: Keywords found in the description (case-insensitive match).
    """
    found = []
    for word in keywords:
        if word.lower() in description.lower():
            found.append(word)
    return found
df["Skills"] = df['Description'].apply(lambda x: comparison(x, skills))
df["Responsibilities"] = df['Description'].apply(lambda x: comparison(x, responsibility))
df["Qualifications"] = df['Description'].apply(lambda x: comparison(x, qualifications))
copy_df = df.copy()
copy_df.drop(columns=["Description"], inplace=True)
copy_df.to_csv("Parsed_dataset.csv", index=False) #Exporting the output for Streamlit
```

The values inside the columns of Job_title, Skills, Responsibilities, Qualifications are going to be split and have their own table along with their IDs and Salary then have their values exploded since it has multiple values containing in every cell

In doing so, it will be easier to present and perform analysis to the dataset with respect to the particular items of each columns

```python
jobTitleTable = copy_df[["ID", "Job_title", "Salary"]].copy()
skillsTable = copy_df[['ID', 'Skills', 'Salary']].copy()
responsibilitiesTable = copy_df[['ID', 'Responsibilities', 'Salary']].copy()
QualificationsTable = copy_df[['ID', 'Qualifications', 'Salary']].copy()
single_Val_Col_Table = copy_df[['ID', 'Date', 'Job_type', 'Hours', 'Salary', 'Link']].copy()
print(jobTitleTable)
print(skillsTable)
print(responsibilitiesTable)
print(QualificationsTable)
```

```python
modified_df = copy_df.copy()
# Removing the [] and ' then split strings by comma into a list in The Job_title Column
jobTitleTable[jobTitleTable.columns[1]] = jobTitleTable[jobTitleTable.columns[1]].str.replace("]", "", regex=False).str.replace("[", "", regex=False).str.replace("'", "", regex=False).str.split(", ")
```
